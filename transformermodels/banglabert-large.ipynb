# Bangla BERT — Automated Essay Scoring (large dataset)

Fine-tunes [`sagorsarker/bangla-bert-base`](https://huggingface.co/sagorsarker/bangla-bert-base)
as a **regression** model that predicts an essay score (0–10) from a
`Question` + `Answer` pair, trained on the **large** dataset (~1,408 graded scripts).

**Pipeline (next cell):**
1. Load and clean the data from `data/essays_large.csv`.
2. Combine question and answer, then tokenize (max length 512).
3. Split into train / validation / test (80 / 10 / 10).
4. Add a linear regression head on top of the `[CLS]` representation.
5. Fine-tune, then report **RMSE**, **MAE**, and **R²** on each split.

> Place the dataset in a local `data/` folder first — see the repo's
> [data/README.md](../data/README.md) for the expected columns.


```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split
import torch
from transformers import AutoTokenizer, Trainer, TrainingArguments, AutoModelForSequenceClassification, AutoModel
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data (place the CSV in a local `data/` folder next to this notebook)
path = os.path.join("data", "essays_large.csv")
tarc_df = pd.read_csv(path)

# Debug: Print the first few rows and check the columns
print(tarc_df.head())  # First 5 rows of the dataset
print(tarc_df.columns)  # Check the column names

# Preprocess the data
df = tarc_df.copy()

# Drop rows where 'Answer' or 'Marks' is missing
df = df[['Question', 'Answer', 'Marks']].dropna(subset=['Answer', 'Marks'])

# Debug: Check how many rows are left after dropping NAs
print(f"Rows after dropping NAs: {len(df)}")

# Ensure 'Marks' column is numeric
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')
df = df.dropna(subset=['Marks'])

# Debug: Verify that the 'Marks' column has been properly converted
print(f"Unique values in 'Marks' column: {df['Marks'].unique()}")

# Combine 'Question' and 'Answer' into one column
df['combined_input'] = "Question: " + df['Question'].astype(str) + " Answer: " + df['Answer'].astype(str)

inputs = df['combined_input'].tolist()
scores = df['Marks'].tolist()

# Debug: Check data types of inputs
print(f"First 5 inputs: {inputs[:5]}")
print(f"Type of first input: {type(inputs[0])}")

# Tokenizer
tokenizer = AutoTokenizer.from_pretrained("sagorsarker/bangla-bert-base")

# Train/test split
train_inputs, temp_inputs, train_scores, temp_scores = train_test_split(inputs, scores, test_size=0.2, random_state=42)
val_inputs, test_inputs, val_scores, test_scores = train_test_split(temp_inputs, temp_scores, test_size=0.5, random_state=42)

# Tokenize with max_length to avoid exceeding sequence length
max_length = 512
train_encodings = tokenizer(train_inputs, padding='max_length', truncation=True, max_length=max_length, return_tensors="pt")
val_encodings = tokenizer(val_inputs, padding='max_length', truncation=True, max_length=max_length, return_tensors="pt")
test_encodings = tokenizer(test_inputs, padding='max_length', truncation=True, max_length=max_length, return_tensors="pt")

# Dataset class
class EssayDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: val[idx].clone().detach() for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx], dtype=torch.float32).clone().detach()
        return item

    def __len__(self):
        return len(self.labels)

# Prepare datasets
train_dataset = EssayDataset(train_encodings, train_scores)
val_dataset = EssayDataset(val_encodings, val_scores)
test_dataset = EssayDataset(test_encodings, test_scores)

# Model setup for regression
class BertForRegression(torch.nn.Module):
    def __init__(self, model_name):
        super(BertForRegression, self).__init__()
        self.bert = AutoModel.from_pretrained(model_name)
        self.regressor = torch.nn.Linear(self.bert.config.hidden_size, 1)

    def forward(self, input_ids, attention_mask, labels=None):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        hidden_state = outputs.last_hidden_state
        pooled_output = hidden_state[:, 0, :]
        logits = self.regressor(pooled_output).view(-1)
        if labels is not None:
            loss_fct = torch.nn.MSELoss()
            loss = loss_fct(logits, labels.view(-1))
            return loss, logits
        return logits

# Training callback to track losses
from transformers import TrainerCallback

class OverfittingCallback(TrainerCallback):
    def __init__(self):
        self.train_losses = []
        self.eval_losses = []

    def on_log(self, args, state, control, logs=None, **kwargs):
        if "loss" in logs:
            self.train_losses.append(logs["loss"])
        if "eval_loss" in logs:
            self.eval_losses.append(logs["eval_loss"])

    def plot_losses(self):
        if not self.train_losses and not self.eval_losses:
            print("No training or validation losses to plot.")
            return
        sns.set(style="whitegrid")
        plt.figure(figsize=(12, 8))
        plt.plot(self.train_losses, label='Training Loss', color='b', linewidth=2, marker='o')
        if self.eval_losses:
            plt.plot(self.eval_losses, label='Validation Loss', color='orange', linewidth=2, linestyle='--', marker='x')
        else:
            print("Warning: No validation losses recorded.")
        plt.xlabel('Steps', fontsize=14)
        plt.ylabel('Loss', fontsize=14)
        plt.title('Training and Validation Loss Over Steps', fontsize=16)
        plt.legend(fontsize=12)
        plt.xticks(fontsize=12)
        plt.yticks(fontsize=12)
        plt.grid(True, alpha=0.6)
        plt.tight_layout()
        plt.show()

# Initialize the callback
overfitting_callback = OverfittingCallback()

# Initialize model
model = BertForRegression("sagorsarker/bangla-bert-base")

# Move model to device (GPU or CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# Training arguments
training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=30,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=32,
    learning_rate=3e-5,
    weight_decay=0.01,
    warmup_steps=1000,
    gradient_accumulation_steps=2,
    lr_scheduler_type="cosine",
    save_steps=1000,
    save_total_limit=3,
    logging_dir='./logs',
    logging_steps=10,
    evaluation_strategy="steps",
    eval_steps=20,
    logging_first_step=True,
    load_best_model_at_end=True,
    max_grad_norm=1.0,
)

# Trainer setup
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset,
    callbacks=[overfitting_callback],
)

# Train the model
trainer.train()

# Plot losses after training
overfitting_callback.plot_losses()

# Evaluate the model
def evaluate_model(dataset, batch_size=32):
    preds, labels = [], []
    for batch in torch.utils.data.DataLoader(dataset, batch_size=batch_size):
        input_ids = batch['input_ids'].to(device)
        attention_mask = batch['attention_mask'].to(device)
        labels_batch = batch['labels'].to(device)
        with torch.no_grad():
            outputs = model(input_ids=input_ids, attention_mask=attention_mask)
        preds.extend(outputs.cpu().numpy())
        labels.extend(labels_batch.cpu().numpy())
    return np.array(preds), np.array(labels)

# Evaluate on training, validation, and test sets
train_preds, train_labels = evaluate_model(train_dataset)
val_preds, val_labels = evaluate_model(val_dataset)
test_preds, test_labels = evaluate_model(test_dataset)

# Calculate and print metrics
def print_metrics(labels, preds, dataset_name):
    mse = mean_squared_error(labels, preds)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(labels, preds)
    r2 = r2_score(labels, preds)
    print(f"{dataset_name} Set Metrics:")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  MAE: {mae:.4f}")
    print(f"  R-squared: {r2:.4f}\n")

print_metrics(train_labels, train_preds, "Training")
print_metrics(val_labels, val_preds, "Validation")
print_metrics(test_labels, test_preds, "Testing")

# Print true and predicted marks for the test set
print("True and Predicted Marks for Test Set:")
for true, pred in zip(test_labels, test_preds):
    print(f"True Marks: {true:.2f}, Predicted Marks: {pred:.2f}")
```

## Visualizing predictions and errors

- **True vs. predicted marks** — points on the dashed diagonal are perfect predictions.
- **Absolute errors** — how far each prediction is from the human score.


```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_true_vs_predictions(true_labels, predictions):
    sns.set(style="whitegrid")
    plt.figure(figsize=(12, 6))
    plt.scatter(true_labels, predictions, alpha=0.6, color="b", label="Predictions")
    plt.plot([min(true_labels), max(true_labels)], [min(true_labels), max(true_labels)],
             color="r", linestyle="--", label="Ideal")
    plt.xlabel("True Marks", fontsize=14)
    plt.ylabel("Predicted Marks", fontsize=14)
    plt.title("True Marks vs Predicted Marks", fontsize=16)
    plt.legend(fontsize=12)
    plt.grid(alpha=0.6)
    plt.tight_layout()
    plt.show()

# Plot True vs Predicted Marks
plot_true_vs_predictions(test_labels, test_preds)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_absolute_errors(true_labels, predictions):
    # Calculate absolute errors
    absolute_errors = [abs(true - pred) for true, pred in zip(true_labels, predictions)]

    sns.set(style="whitegrid")
    plt.figure(figsize=(12, 6))
    plt.scatter(range(len(absolute_errors)), absolute_errors, color="orange", alpha=0.7, label="Absolute Errors")
    plt.axhline(y=0, color="r", linestyle="--", label="No Error")
    plt.xlabel("Sample Index", fontsize=14)
    plt.ylabel("Absolute Error", fontsize=14)
    plt.title("Absolute Errors in Predictions", fontsize=16)
    plt.legend(fontsize=12)
    plt.grid(alpha=0.6)
    plt.tight_layout()
    plt.show()

# Call the function with true and predicted marks
plot_absolute_errors(test_labels, test_preds)

```
