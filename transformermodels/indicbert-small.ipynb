# IndicBERT (AI4Bharat) — Automated Essay Scoring (small dataset)

Fine-tunes [`ai4bharat/indic-bert`](https://huggingface.co/ai4bharat/indic-bert)
as a **regression** model that predicts an essay score (0–10) from a
`Question` + `Answer` pair, trained on the **small** dataset (~325 graded scripts).

**Pipeline (next cell):**
1. Load and clean the data from `data/essays_small.csv`.
2. Build a Hugging Face `Dataset`, then tokenize (max length 512).
3. Split into train / validation / test (80 / 10 / 10).
4. Add a linear regression head on top of the `[CLS]` representation.
5. Fine-tune with early stopping, then report **RMSE**, **MAE**, and **R²**.

> Place the dataset in a local `data/` folder first — see the repo's
> [data/README.md](../data/README.md) for the expected columns.


```python
# Optional: install dependencies (or use the repo's requirements.txt)
%pip install -q transformers datasets pyarrow matplotlib

# Import necessary libraries
import pandas as pd
import torch
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from transformers import AutoTokenizer, TrainingArguments, AutoModel
from datasets import Dataset
from transformers import TrainerCallback, Trainer, EarlyStoppingCallback
import os
import matplotlib.pyplot as plt

# Function to make tensors contiguous
def make_tensors_contiguous(state_dict):
    return {k: v.contiguous() for k, v in state_dict.items()}

# CustomTrainer class with modified _save method
class CustomTrainer(Trainer):
    def _save(self, output_dir: str, state_dict=None):
        if state_dict is None:
            state_dict = make_tensors_contiguous(self.model.state_dict())
        torch.save(state_dict, os.path.join(output_dir, "pytorch_model.bin"))

# Load the dataset (place the CSV in a local `data/` folder next to this notebook)
path = os.path.join("data", "essays_small.csv")
tarc_df = pd.read_csv(path)

# Debug: Print the first few rows and check the columns
print(tarc_df.head())  # First 5 rows of the dataset
print(tarc_df.columns)  # Check the column names

df = tarc_df.copy()

# Drop rows where 'Answer' or 'Marks' is missing
df = df[['Question', 'Answer', 'Marks']].dropna(subset=['Answer', 'Marks'])

# Ensure 'Marks' column is numeric
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')
df = df.dropna(subset=['Marks'])

# Combine 'Question' and 'Answer' into one column
df['essays'] = "Question: " + df['Question'] + " Answer: " + df['Answer']
df['scores'] = df['Marks']

# Convert to Hugging Face Dataset
dataset = Dataset.from_pandas(df[['essays', 'scores']])

# Split dataset into train, validation, and test sets
split_dataset = dataset.train_test_split(test_size=0.2)
train_dataset = split_dataset['train']
remaining_dataset = split_dataset['test']

# Further split the remaining dataset into validation and test sets
split_remaining = remaining_dataset.train_test_split(test_size=0.5)
val_dataset = split_remaining['train']
test_dataset = split_remaining['test']

# Define tokenizer and model
model_name = "ai4bharat/indic-bert"
tokenizer = AutoTokenizer.from_pretrained(model_name, use_fast=False, trust_remote_code=True)

# Tokenize the dataset
def tokenize_function(examples):
    return tokenizer(examples['essays'], truncation=True, padding='max_length', max_length=512)

# Tokenize datasets
tokenized_train = train_dataset.map(tokenize_function, batched=True)
tokenized_val = val_dataset.map(tokenize_function, batched=True)
tokenized_test = test_dataset.map(tokenize_function, batched=True)

# Rename 'scores' to 'labels' as expected by Trainer
tokenized_train = tokenized_train.rename_column("scores", "labels")
tokenized_val = tokenized_val.rename_column("scores", "labels")
tokenized_test = tokenized_test.rename_column("scores", "labels")

# Set PyTorch format
tokenized_train.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])
tokenized_val.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])
tokenized_test.set_format(type="torch", columns=["input_ids", "attention_mask", "labels"])

# Model setup for regression
class BertForRegression(torch.nn.Module):
    def __init__(self, model_name):
        super(BertForRegression, self).__init__()
        self.bert = AutoModel.from_pretrained(model_name)
        self.regressor = torch.nn.Linear(self.bert.config.hidden_size, 1)  # Single output for regression

    def forward(self, input_ids, attention_mask, labels=None):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        pooled_output = outputs.last_hidden_state[:, 0, :].contiguous()
        logits = self.regressor(pooled_output).view(-1).contiguous()
        if labels is not None:
            loss_fct = torch.nn.MSELoss()
            loss = loss_fct(logits, labels.view(-1))
            return loss, logits
        return logits

# Callback for tracking overfitting
class OverfittingCallback(TrainerCallback):
    def __init__(self):
        self.train_losses = []
        self.eval_losses = []

    def on_log(self, args, state, control, logs=None, **kwargs):
        if "loss" in logs:
            self.train_losses.append(logs["loss"])
        if "eval_loss" in logs:
            self.eval_losses.append(logs["eval_loss"])

    def plot_losses(self):
        if self.train_losses and self.eval_losses:
            plt.figure(figsize=(10, 6))
            plt.plot(self.train_losses, label="Training Loss")
            plt.plot(self.eval_losses, label="Validation Loss")
            plt.title("Training vs. Validation Loss")
            plt.xlabel("Steps")
            plt.ylabel("Loss")
            plt.legend()
            plt.grid()
            plt.show()

# Initialize the callback
overfitting_callback = OverfittingCallback()

# Initialize model
model = BertForRegression(model_name)

# Move model to device (GPU or CPU)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)

# Metric computation
def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = predictions.squeeze()
    mse = mean_squared_error(labels, predictions)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(labels, predictions)
    r2 = r2_score(labels, predictions)
    return {
        "eval_mse": mse,
        "eval_rmse": rmse,
        "eval_mae": mae,
        "eval_r2": r2,
    }

# Training arguments
training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=30,
    per_device_train_batch_size=2,
    per_device_eval_batch_size=16,
    learning_rate=3e-5,
    weight_decay=0.01,
    warmup_steps=1000,
    gradient_accumulation_steps=4,
    lr_scheduler_type="cosine",
    save_steps=1000,
    save_total_limit=3,
    logging_dir="./logs",
    logging_steps=50,
    evaluation_strategy="steps",
    eval_steps=20,
    logging_first_step=True,
    load_best_model_at_end=True,
    metric_for_best_model="eval_rmse",
    greater_is_better=False,
    max_grad_norm=1.0,
)

# Define EarlyStoppingCallback
early_stopping = EarlyStoppingCallback(
    early_stopping_patience=3,  # Stop after 3 evaluations without improvement
    early_stopping_threshold=0.001  # Minimum change in metric to qualify as improvement
)

# Trainer initialization
trainer = CustomTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_val,
    compute_metrics=compute_metrics,
    callbacks=[overfitting_callback, early_stopping],  # Include early stopping
)

# Train the model
trainer.train()

# Evaluate on validation dataset
results = trainer.predict(tokenized_val)

# Extract predictions and labels
predictions = results.predictions.squeeze()
labels = results.label_ids

# Print true vs. predicted marks
print("True Marks vs. Predicted Marks:")
for true, pred in zip(labels, predictions):
    print(f"True: {true:.2f}, Predicted: {pred:.2f}")

# Plot training vs. validation loss
overfitting_callback.plot_losses()

# Evaluate on train, validation, and test datasets
train_results = trainer.predict(tokenized_train)
val_results = trainer.predict(tokenized_val)
test_results = trainer.predict(tokenized_test)

# Helper function to calculate metrics
def evaluate_results(results, dataset_name):
    predictions = results.predictions.squeeze()
    labels = results.label_ids
    mse = mean_squared_error(labels, predictions)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(labels, predictions)
    r2 = r2_score(labels, predictions)
    print(f"{dataset_name} Set Metrics:")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  MAE: {mae:.4f}")
    print(f"  R-squared: {r2:.4f}\n")

# Print metrics for train, validation, and test sets
evaluate_results(train_results, "Training")
evaluate_results(val_results, "Validation")
evaluate_results(test_results, "Testing")
```

## Visualizing predictions and errors

Diagnostic plots for each split: true vs. predicted marks, residuals, a sorted
line comparison, and per-sample absolute errors.


```python
# Plot training vs validation loss
def plot_training_validation_loss(train_losses, eval_losses):
    plt.figure(figsize=(10, 6))
    plt.plot(train_losses, label="Training Loss", color="blue", linewidth=2)
    plt.plot(eval_losses, label="Validation Loss", color="orange", linestyle="--", linewidth=2)
    plt.title("Training vs Validation Loss Over Steps", fontsize=16)
    plt.xlabel("Steps", fontsize=14)
    plt.ylabel("Loss", fontsize=14)
    plt.legend(fontsize=12)
    plt.grid(alpha=0.3)
    plt.show()

# Plot true vs predicted values
def plot_true_vs_predicted(labels, predictions, dataset_name):
    plt.figure(figsize=(8, 6))
    plt.scatter(labels, predictions, alpha=0.6, color="blue", label="Predictions")
    plt.plot([min(labels), max(labels)], [min(labels), max(labels)], color="red", linestyle="--", label="Ideal Fit")
    plt.title(f"True vs Predicted ({dataset_name} Set)", fontsize=16)
    plt.xlabel("True Values", fontsize=14)
    plt.ylabel("Predicted Values", fontsize=14)
    plt.legend(fontsize=12)
    plt.grid(alpha=0.3)
    plt.show()

# Residual plot
def plot_residuals(labels, predictions, dataset_name):
    residuals = labels - predictions
    plt.figure(figsize=(8, 6))
    plt.scatter(predictions, residuals, alpha=0.6, color="purple")
    plt.axhline(y=0, color="red", linestyle="--", linewidth=2)
    plt.title(f"Residual Plot ({dataset_name} Set)", fontsize=16)
    plt.xlabel("Predicted Values", fontsize=14)
    plt.ylabel("Residuals", fontsize=14)
    plt.grid(alpha=0.3)
    plt.show()

# True vs Predicted Line Plot
def plot_true_vs_predicted_line(labels, predictions, dataset_name):
    indices = np.argsort(labels)
    sorted_labels = np.array(labels)[indices]
    sorted_predictions = np.array(predictions)[indices]
    
    plt.figure(figsize=(10, 6))
    plt.plot(sorted_labels, label="True Values", color="blue", linewidth=2)
    plt.plot(sorted_predictions, label="Predicted Values", color="orange", linestyle="--", linewidth=2)
    plt.title(f"True vs Predicted Line Plot ({dataset_name} Set)", fontsize=16)
    plt.xlabel("Index", fontsize=14)
    plt.ylabel("Values", fontsize=14)
    plt.legend(fontsize=12)
    plt.grid(alpha=0.3)
    plt.show()

# Absolute Error Plot
def plot_absolute_errors(labels, predictions, dataset_name):
    absolute_errors = np.abs(labels - predictions)
    plt.figure(figsize=(10, 6))
    plt.bar(range(len(absolute_errors)), absolute_errors, color="cyan", edgecolor="black")
    plt.title(f"Absolute Errors for Each Prediction ({dataset_name} Set)", fontsize=16)
    plt.xlabel("Index", fontsize=14)
    plt.ylabel("Absolute Error", fontsize=14)
    plt.grid(alpha=0.3)
    plt.show()

# Evaluate and plot for train, validation, and test datasets
def evaluate_and_plot(results, dataset_name):
    predictions = results.predictions.squeeze()
    labels = results.label_ids
    mse = mean_squared_error(labels, predictions)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(labels, predictions)
    r2 = r2_score(labels, predictions)
    print(f"{dataset_name} Set Metrics:")
    print(f"  RMSE: {rmse:.4f}")
    print(f"  MAE: {mae:.4f}")
    print(f"  R-squared: {r2:.4f}\n")

    # Generate plots
    plot_true_vs_predicted(labels, predictions, dataset_name)
    plot_residuals(labels, predictions, dataset_name)
    plot_true_vs_predicted_line(labels, predictions, dataset_name)
    plot_absolute_errors(labels, predictions, dataset_name)

# Call plotting functions for train, validation, and test datasets
evaluate_and_plot(train_results, "Training")
evaluate_and_plot(val_results, "Validation")
evaluate_and_plot(test_results, "Testing")
```
